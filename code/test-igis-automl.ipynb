## 依赖

```python
from autogluon.tabular import TabularDataset, TabularPredictor
```

## 构建训练数据集

```python
# data_url = 'https://raw.githubusercontent.com/mli/ag-docs/main/knot_theory/'
# train_data = TabularDataset(f'{data_url}train.csv')
# test_data = TabularDataset(f'{data_url}test.csv')

train_data = TabularDataset('./dataset/aws/train.csv')

print(train_data.head())
```

## 设置标签列

```python
label = 'class'
print(train_data[label].describe())
```

## 训练模型

```python
save_path = 'models/aws'
predictor = TabularPredictor(label=label, path=save_path).fit(train_data)

# 加载模型（加载一个已训练好的模型）
# predictor = TabularPredictor.load('./models/aws')
```

## 构建测试数据集

```python
test_data = TabularDataset('./dataset/aws/test.csv')
print(test_data.head())
```

## 分离测试集真值

```python
y_true = test_data[label]
test_data_nolab = test_data.drop(columns=[label])
print(test_data_nolab.head())
```

## 模型推理d)

```python
y_pred = predictor.predict(test_data_nolab)
print(y_pred)
```

## 手动模型评估

```python
pref = predictor.evaluate_predictions(y_true=y_true, y_pred=y_pred)
print(pref)
```

## 自动模型评估

```python
prefs = predictor.leaderboard(test_data, silent=True)
print(prefs)
```

## 多分类评估（计算各类型概率）

```python
pred_probs = predictor.predict_proba(test_data_nolab)
print(pred_probs.head())
```

## 模型指标（模型在训练集上的表现）

```python
predictor_info = predictor.fit_summary(show_plot=True)
print(predictor_info)
```

## 模型类型信息

```python
print("AutoGluon infers problem type is: ", predictor.problem_type)
```

## 模型可推导的信息（数据列）

```python
print(predictor.feature_metadata)
```
